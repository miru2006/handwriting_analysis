```python

# HANDWRITTEN CHARACTER RECOGNITION USING CNN

# 1. IMPORT LIBRARIES

import pandas as pd
import numpy as np
import tensorflow as tf
import matplotlib.pyplot as plt
import cv2

from google.colab import files
from PIL import Image


# 2. UPLOAD EMNIST DATASET




# 3. LOAD DATASET

train = pd.read_csv("emnist-letters-train.csv", header=None)
test = pd.read_csv("emnist-letters-test.csv", header=None)

if train.shape[1] == 1:
    train = pd.read_csv("emnist-letters-train.csv",
                        sep="\t", header=None)

if test.shape[1] == 1:
    test = pd.read_csv("emnist-letters-test.csv",
                       sep="\t", header=None)

if train.shape[1] != 785 or test.shape[1] != 785:
    raise ValueError("Incorrect dataset format!")

print("Training data:", train.shape)
print("Testing data:", test.shape)


# 4. SEPARATE LABELS AND IMAGES

X_train = train.iloc[:, 1:].values
y_train = train.iloc[:, 0].values

X_test = test.iloc[:, 1:].values
y_test = test.iloc[:, 0].values


# 5. PREPROCESS DATASET

X_train = X_train.reshape(-1, 28, 28, 1)
X_test = X_test.reshape(-1, 28, 28, 1)

X_train = X_train.astype("float32") / 255.0
X_test = X_test.astype("float32") / 255.0

y_train = y_train.astype("int32") - 1
y_test = y_test.astype("int32") - 1


# DISPLAY TRAINING SAMPLES

plt.figure(figsize=(10, 4))

for i in range(10):
    plt.subplot(2, 5, i + 1)
    plt.imshow(X_train[i].reshape(28, 28), cmap="gray")
    plt.title(chr(y_train[i] + 65))
    plt.axis("off")

plt.suptitle("EMNIST Training Samples")
plt.show()


# 6. BUILD IMPROVED CNN MODEL

model = tf.keras.Sequential([

    tf.keras.layers.Input(shape=(28, 28, 1)),

    # Data augmentation
    tf.keras.layers.RandomRotation(0.04),
    tf.keras.layers.RandomTranslation(0.08, 0.08),
    tf.keras.layers.RandomZoom(0.08),

    # First convolution block
    tf.keras.layers.Conv2D(32, (3, 3), padding="same",
                           activation="relu"),
    tf.keras.layers.BatchNormalization(),
    tf.keras.layers.MaxPooling2D((2, 2)),

    # Second convolution block
    tf.keras.layers.Conv2D(64, (3, 3), padding="same",
                           activation="relu"),
    tf.keras.layers.BatchNormalization(),
    tf.keras.layers.MaxPooling2D((2, 2)),

    # Third convolution block
    tf.keras.layers.Conv2D(128, (3, 3), padding="same",
                           activation="relu"),
    tf.keras.layers.BatchNormalization(),

    tf.keras.layers.Flatten(),

    tf.keras.layers.Dense(256, activation="relu"),
    tf.keras.layers.Dropout(0.4),

    tf.keras.layers.Dense(26, activation="softmax")
])


# 7. COMPILE MODEL

model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

model.summary()


# 8. TRAIN MODEL

early_stop = tf.keras.callbacks.EarlyStopping(
    monitor="val_accuracy",
    patience=4,
    restore_best_weights=True
)

history = model.fit(
    X_train,
    y_train,
    epochs=20,
    batch_size=128,
    validation_split=0.1,
    callbacks=[early_stop]
)


# 9. EVALUATE MODEL

loss, accuracy = model.evaluate(X_test, y_test)

print("\nTest Accuracy:", round(accuracy * 100, 2), "%")


# 10. DISPLAY ACCURACY GRAPH

plt.figure(figsize=(8, 5))

plt.plot(history.history["accuracy"],
         label="Training Accuracy")

plt.plot(history.history["val_accuracy"],
         label="Validation Accuracy")

plt.xlabel("Epochs")
plt.ylabel("Accuracy")
plt.title("CNN Training Accuracy")
plt.legend()
plt.show()


# 11. SAVE MODEL

model.save("handwritten_character_model.keras")

print("Model saved successfully!")


# 12. HANDWRITTEN IMAGE PREPROCESSING FUNCTION

def preprocess_handwriting(image_path):

    # Read image in grayscale
    img = cv2.imread(image_path, cv2.IMREAD_GRAYSCALE)

    if img is None:
        raise ValueError("Unable to read image!")

    # Check background brightness
    border = np.concatenate([
        img[0, :], img[-1, :],
        img[:, 0], img[:, -1]
    ])

    # Convert handwriting into white strokes
    # on a black background
    if np.mean(border) > 127:
        img = 255 - img

    # Apply Otsu thresholding
    _, binary = cv2.threshold(
        img, 0, 255,
        cv2.THRESH_BINARY + cv2.THRESH_OTSU
    )

    # Find character boundaries
    rows, cols = np.where(binary > 0)

    if len(rows) == 0:
        raise ValueError("No character detected!")

    # Crop character
    binary = binary[
        rows.min():rows.max() + 1,
        cols.min():cols.max() + 1
    ]

    # Resize while maintaining aspect ratio
    h, w = binary.shape

    scale = 20 / max(h, w)

    new_w = max(1, int(w * scale))
    new_h = max(1, int(h * scale))

    resized = cv2.resize(
        binary,
        (new_w, new_h),
        interpolation=cv2.INTER_AREA
    )

    # Center inside 28x28 canvas
    canvas = np.zeros((28, 28), dtype=np.uint8)

    x = (28 - new_w) // 2
    y = (28 - new_h) // 2

    canvas[y:y + new_h, x:x + new_w] = resized

    # Use the 270-degree orientation
    # based on the previous orientation test
    canvas = np.rot90(canvas, 3)

    # Normalize
    canvas = canvas.astype("float32") / 255.0

    return canvas


# 13. UPLOAD YOUR HANDWRITTEN CHARACTER

print("\nUpload one handwritten character image (JPG or PNG)")

uploaded_image = files.upload()

image_path = list(uploaded_image.keys())[0]


# 14. PREPROCESS UPLOADED IMAGE

processed_image = preprocess_handwriting(image_path)

plt.figure(figsize=(4, 4))
plt.imshow(processed_image, cmap="gray")
plt.title("Image Given to CNN")
plt.axis("off")
plt.show()


# 15. PREDICT CHARACTER

prediction = model.predict(
    processed_image.reshape(1, 28, 28, 1),
    verbose=0
)[0]

top_indices = np.argsort(prediction)[-3:][::-1]

predicted_character = chr(top_indices[0] + 65)
confidence = prediction[top_indices[0]] * 100


# 16. DISPLAY TOP THREE PREDICTIONS

print("\nPREDICTION RESULTS")

for index in top_indices:

    character = chr(index + 65)
    probability = prediction[index] * 100

    print(character, ":", round(probability, 2), "%")


print("\nPredicted Character:", predicted_character)
print("Confidence:", round(confidence, 2), "%")


# 17. DISPLAY FINAL RESULT

plt.figure(figsize=(4, 4))

plt.imshow(processed_image, cmap="gray")

plt.title(
    "Predicted Character: " + predicted_character
)

plt.axis("off")
plt.show()
```
